# 1. Estimating cetane number for diesel fuel

## 1.1 Load datasets
Importing, loading datasets and printing datasets

```python
%matplotlib notebook
%matplotlib inline
import matplotlib.pylab as plt
import numpy as np
from sklearn.decomposition import PCA
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from scipy.io import loadmat

# Loading the datasets
mat = loadmat("datasets/cnDieselTrain.mat")
X_train = mat["cnTrainX"].T
y_train = mat["cnTrainY"].reshape(len(X_train))
X_test = mat["cnTestX"].T

print("X_train shape: {}".format(X_train.shape))
print("y_train shape: {}".format(y_train.shape))
print("X_test shape: {}".format(X_test.shape))
print("\n\033[1mX_train[:5, :5]:\033[0m\n{}".format(X_train[:5, :5]))
print("\n\033[1my_train[:5]:\033[0m\n{}".format(y_train[:5]))
```

## 1.2 PCA, features, importance, visualizing
Decide the amount/which features are to be used. This is based on an analysis done on the cummulative variance and the variance ratio from the features. The cummulated variance is plotted and so is the two most "important" features in order to visualize the data to a certain degree for us. After looking at what we got, we decided that <b>n_components</b> should be 8 when using PCA. This is because at the <b>8th component</b> the cummulative variance is greater than 99% and it's the last component where the variance is greater than 0.1%.

```python
pca = PCA().fit(X_train)
fig1, ax1=plt.subplots()
ax1.plot(np.cumsum(pca.explained_variance_ratio_)[:20])
print("\n\033[1mCummulative variance of the top [0..10]:\033[0m \n{}".format(np.cumsum(pca.explained_variance_ratio_)[:10]))
print("\n\033[1mVariance in the [0..10] top features:\033[0m \n{}".format(pca.explained_variance_ratio_[:10]))
ax1.set_xlabel("number of features")
ax1.set_ylabel("cummulative explained variance")

X_train_pca = pca.transform(X_train)
fig2, ax2 = plt.subplots()
ax2.scatter(X_train_pca[:,0], X_train_pca[:,1],marker=".")
ax2.set_xlabel("First principal componant of PCA")
ax2.set_ylabel("Second principal componant of PCA")

#print("pca shape:", X_train_pca.shape)
#print(X_train_pca[:5, :5])
""" 
After looking at the features in the matrix returned by PCA we can see that the data is in need of being normalized.
Let's see how it looks when plotted.
"""
#fig, ax = plt.subplots()
#ax.scatter(X_train_pca[:, 0], X_train_pca[:, 1], marker=".")


#scaler = StandardScaler()
#scaler.fit(X_train,y=y_train)  # doctest: +SKIP
#X_train = scaler.transform(X_train)  # doctest: +SKIP
# apply same transformation to test data
#X_test = scaler.transform(X_test)  # doctest: +SKIP

#print("Training MLPRegressor...")
#est = MLPRegressor(solver='lbfgs', learning_rate_init=0.001, max_iter=500, alpha=0.001)
#est.fit(X_train, y_train)
#print("Test R2 score: {:.2f}".format(est.score(X_train, y_train)))
#predicted = est.predict(X_train)

#print("Cost: ",sum((predicted-y_train)**2)/(2*len(y_train)))
#print(y_train)
```
